## Question Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
```python
Input: nums = [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```
```python
Input: nums = [4, 0, 5, 0, 0, 6]
Output: [4, 5, 6, 0, 0, 0]
```


````python
from typing import List # Allows a function to use data in the form of a list

nums = [0, 1, 0, 3, 12]     # Copying the list from the question prompt

def move_zeros_to_end(nums: List[int]) -> List[int]:     # Starting with a list of integers and returning a list of integers
    index = 0       # Start checking at position 0 in the list 'nums'
    checks_remaining = len(nums)    # Continue the process for the legnth of the list called 'nums'

    while checks_remaining > 0:     # if there are more integers remaining in the list, continue and only check the original amount of integers
        if nums[index] == 0:        # check if the first positioned integer in the list is equal to 0
            nums.pop(index)         # if equal to 0 remove (pop) the integer from the start of the list
            nums.append(0)          # add (append) the 0 to the end of the list
        else:
            index += 1

        checks_remaining -= 1   # if the value of the integer in the list is not equal to 0, keep moving down the index numbers in the list

    return nums

print(move_zeros_to_end(nums))
````

### Answer

-   Explain the problem’s time and space complexity


### Answer

Regarding my original solution to Part 2:

+ the time complexity is (at worst) O(n^2). This is because the loop first runs for each integer in the list (n) and then decides whether to pop the value of the integer based on its value (n) and then append the value to the end if it is equal to 0. So, O(n)*O(n)=O(n^2)

+ the space complexity is O(1) because it works from the original list directly, defined my the total number of integers, and only makes new variables (not new lists)
